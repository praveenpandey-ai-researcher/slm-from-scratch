```python
# CELL 4: Collect 1 Million GPT-4 Rows
# ======================================
import time

TARGET        = 1_000_000
GPT4_PREFIXES = ("niv.", "flan.")

print("="*55)
print(f"  COLLECTING {TARGET:,} GPT-4 ROWS")
print("="*55)
print()
print("Filter: keep ids starting with 'niv.' or 'flan.'")
print("Progress every 100,000 rows. ~10-15 minutes.")
print("-"*55)

gpt4_samples  = []
total_seen    = 0
total_skipped = 0
last_milestone= 0
start         = time.time()

for row in dataset:
    total_seen += 1
    row_id      = row.get("id", "")

    if row_id.startswith(GPT4_PREFIXES):
        gpt4_samples.append(row)
    else:
        total_skipped += 1

    # Print once per 100k milestone
    current = len(gpt4_samples)
    if current >= last_milestone + 100_000:
        last_milestone = current
        pct     = current / TARGET * 100
        elapsed = time.time() - start
        eta     = (elapsed / max(current, 1)) * (TARGET - current)
        bar     = "█"*int(pct//5) + "░"*(20-int(pct//5))
        print(f"  [{bar}] {pct:5.1f}% | "
              f"GPT-4: {current:>9,} | "
              f"Seen: {total_seen:>9,} | "
              f"ETA: {eta/60:.1f}min")

    if current >= TARGET:
        break

elapsed = time.time() - start

print()
print("="*55)
print("🎉 COLLECTION COMPLETE!")
print("="*55)
print(f"  GPT-4 rows collected : {len(gpt4_samples):,}")
print(f"  GPT-3.5 rows skipped : {total_skipped:,}")
print(f"  Total rows scanned   : {total_seen:,}")
print(f"  Time taken           : {elapsed/60:.1f} minutes")
print()

ex = gpt4_samples[0]
print("📝 Sample GPT-4 row:")
print("-"*55)
print(f"  id       : {ex['id']}")
print(f"  question : {ex['question'][:80]}...")
print(f"  response : {ex['response'][:120]}...")
print()
print("✅ CELL 4 COMPLETE — 1M GPT-4 rows in memory!")
print("   Next → CELL 5: Explore the data")
print("="*55)
```

```python
# CELL 7: Format & Save Training File
# =====================================
# Converts raw data → clean format → saves to disk
# Shows progress every 50,000 rows in real-time

import os
import time

OUTPUT = "/kaggle/working/training_data.txt"

# --------------------------------------------------
# Formatter function
# --------------------------------------------------
def format_example(ex):
    """Convert one raw row → clean training text."""
    system   = ex['system_prompt'].strip()
    question = ex['question'].strip()
    response = ex['response'].strip()

    # Quality filters
    if len(response) < 30:   return None  # too short
    if len(response) > 2000: return None  # too long
    if len(question) < 5:    return None  # garbage question

    # Default system prompt if empty
    if not system:
        system = "You are a helpful AI assistant."

    return (
        f"### System:\n{system}\n\n"
        f"### Question:\n{question}\n\n"
        f"### Answer:\n{response}\n\n"
        f"{'='*40}\n\n"
    )

# --------------------------------------------------
# Quick sanity test
# --------------------------------------------------
print("="*55)
print("  STEP 1: Testing formatter")
print("="*55)
test = format_example(gpt4_samples[0])
print(f"  Test row formatted: {'✅ OK' if test else '❌ filtered'}")
print()

# --------------------------------------------------
# Format all rows with real-time progress
# --------------------------------------------------
print("="*55)
print("  STEP 2: Formatting 1,000,000 rows")
print("="*55)
print()
print("  Progress every 50,000 rows:")
print("  " + "-"*50)

all_parts = []
included  = 0
skipped   = 0
start     = time.time()

for i, ex in enumerate(gpt4_samples):
    result = format_example(ex)
    if result:
        all_parts.append(result)
        included += 1
    else:
        skipped += 1

    # Real-time progress every 50k
    if (i + 1) % 50_000 == 0:
        pct     = (i + 1) / len(gpt4_samples) * 100
        elapsed = time.time() - start
        eta     = (elapsed / (i+1)) * (len(gpt4_samples) - i - 1)
        bar     = "█"*int(pct//5) + "░"*(20-int(pct//5))
        print(f"  [{bar}] {pct:5.1f}% | "
              f"Kept: {included:>7,} | "
              f"Skip: {skipped:>6,} | "
              f"ETA: {eta:.0f}s")

print()
print(f"  ✅ Formatting done in {time.time()-start:.1f}s")
print(f"     Kept: {included:,} | Skipped: {skipped:,}")
print()

# --------------------------------------------------
# Combine identity + OpenOrca, write to disk
# --------------------------------------------------
print("="*55)
print("  STEP 3: Writing to disk")
print("="*55)
print()

write_start = time.time()
full_text   = identity_data + "".join(all_parts)

print(f"  Combining identity + OpenOrca...")
print(f"  Total size in memory: {len(full_text)/1e6:.0f} MB")
print(f"  Writing to {OUTPUT}...")

with open(OUTPUT, "w", encoding="utf-8") as f:
    f.write(full_text)

print(f"  ✅ Written in {time.time()-write_start:.1f}s")
print()

# --------------------------------------------------
# Final summary
# --------------------------------------------------
file_mb = os.path.getsize(OUTPUT) / 1e6
total_qa= full_text.count("### Question:")

print("="*55)
print("🎉 TRAINING FILE SAVED!")
print("="*55)
print(f"  Location      : {OUTPUT}")
print(f"  File size     : {file_mb:.0f} MB")
print(f"  OpenOrca pairs: {included:,}")
print(f"  Identity pairs: {identity_data.count('### Question:')}")
print(f"  Total Q&A     : {total_qa:,}")
print(f"  Total time    : {time.time()-start:.1f}s")
print()
print("  Next → CELL 8: Tokenization")
print("="*55)
```

```python
# CELL 8: Tokenization
# ======================
# Convert training_data.txt (text) → tokens.bin (numbers)
# Shows progress for every chunk in real-time

import os
import numpy as np
import time
from transformers import GPT2TokenizerFast

# --------------------------------------------------
# STEP 1: Load tokenizer
# --------------------------------------------------
print("="*55)
print("  STEP 1: Load GPT-2 Tokenizer")
print("="*55)
print()
print("📥 Loading tokenizer...")

tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")

print(f"✅ Loaded! Vocabulary: {tokenizer.vocab_size:,} tokens")
print()

# --------------------------------------------------
# STEP 2: Show how it works
# --------------------------------------------------
print("="*55)
print("  STEP 2: Tokenizer Examples")
print("="*55)
print()

for text in ["Who are you?", "### Question:", "### Answer:"]:
    toks = tokenizer.encode(text)
    print(f"  '{text}'")
    print(f"   → {toks}  ({len(toks)} tokens)")
    print()

# --------------------------------------------------
# STEP 3: Tokenize the full file in chunks
# --------------------------------------------------
INPUT_FILE  = "/kaggle/working/training_data.txt"
OUTPUT_FILE = "/kaggle/working/tokens.bin"
CHUNK_CHARS = 1_000_000   # 1MB of text per chunk

print("="*55)
print("  STEP 3: Tokenizing Full File")
print("="*55)
print()

file_size    = os.path.getsize(INPUT_FILE)
total_tokens = 0
chunk_num    = 0
start        = time.time()

print(f"  Input  : {INPUT_FILE}")
print(f"  Output : {OUTPUT_FILE}")
print(f"  Size   : {file_size/1e6:.0f} MB to process")
print(f"  Chunks : ~{file_size//CHUNK_CHARS + 1} chunks")
print()
print("  Progress (every chunk):")
print("  " + "-"*50)

with open(OUTPUT_FILE, "wb") as out_f:
    with open(INPUT_FILE, "r", encoding="utf-8") as in_f:

        while True:
            chunk_text = in_f.read(CHUNK_CHARS)
            if not chunk_text:
                break

            chunk_num += 1

            # Tokenize this chunk
            token_ids = tokenizer.encode(chunk_text)

            # Save as uint16 (2 bytes per token)
            arr = np.array(token_ids, dtype=np.uint16)
            arr.tofile(out_f)

            total_tokens += len(token_ids)

            # Real-time progress
            bytes_done = in_f.tell()
            pct        = bytes_done / file_size * 100
            elapsed    = time.time() - start
            eta        = (elapsed/max(bytes_done,1)) * \
                         (file_size - bytes_done)
            bar        = "█"*int(pct//5) + "░"*(20-int(pct//5))

            print(f"  Chunk {chunk_num:>4} | "
                  f"[{bar}] {pct:5.1f}% | "
                  f"Tokens: {total_tokens:>12,} | "
                  f"ETA: {eta/60:.1f}min")

elapsed = time.time() - start

# --------------------------------------------------
# STEP 4: Verify & summary
# --------------------------------------------------
tokens  = np.memmap(OUTPUT_FILE, dtype=np.uint16, mode='r')
file_mb = os.path.getsize(OUTPUT_FILE) / 1e6

print()
print("="*55)
print("🎉 TOKENIZATION COMPLETE!")
print("="*55)
print(f"  Total chunks    : {chunk_num:,}")
print(f"  Total tokens    : {len(tokens):,}")
print(f"  Token file size : {file_mb:.0f} MB")
print(f"  Time taken      : {elapsed/60:.1f} minutes")
print(f"  Speed           : {len(tokens)/elapsed:,.0f} tokens/sec")
print()
print(f"  First 15 tokens : {tokens[:15].tolist()}")
print(f"  Decodes to      : '{tokenizer.decode(tokens[:15].tolist())}'")
print()
print("  Next → CELL 9: Build the Model 🧠")
print("="*55)
```

```python
# CELL 9: Build the Transformer Model
# =====================================
# Build every component from scratch.

import torch
import torch.nn as nn
import torch.nn.functional as F
import math

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# --------------------------------------------------
# CONFIGURATION
# --------------------------------------------------
class ModelConfig:
    vocab_size = 50257   # GPT-2 vocabulary
    n_embd     = 384     # embedding dimensions
    n_heads    = 6       # attention heads
    n_layers   = 6       # transformer blocks
    block_size = 256     # context window
    dropout    = 0.1     # regularization

config = ModelConfig()

print("="*55)
print("  MODEL CONFIGURATION")
print("="*55)
print(f"  vocab_size : {config.vocab_size:,}")
print(f"  n_embd     : {config.n_embd}")
print(f"  n_heads    : {config.n_heads}")
print(f"  n_layers   : {config.n_layers}")
print(f"  block_size : {config.block_size}")
print()

# --------------------------------------------------
# COMPONENT 1: Self Attention
# --------------------------------------------------
class SelfAttention(nn.Module):
    """Multi-head self attention — the core of Transformer."""
    def __init__(self, config):
        super().__init__()
        assert config.n_embd % config.n_heads == 0
        self.n_heads   = config.n_heads
        self.n_embd    = config.n_embd
        self.head_size = config.n_embd // config.n_heads
        # Combined QKV projection (faster than 3 separate)
        self.qkv_proj  = nn.Linear(config.n_embd, 3*config.n_embd, bias=False)
        self.out_proj  = nn.Linear(config.n_embd, config.n_embd, bias=False)
        self.dropout   = nn.Dropout(config.dropout)
        # Causal mask — prevents looking at future tokens
        self.register_buffer(
            "mask",
            torch.tril(torch.ones(config.block_size, config.block_size))
            .view(1, 1, config.block_size, config.block_size)
        )

    def forward(self, x):
        B, T, C = x.shape
        qkv     = self.qkv_proj(x)
        Q, K, V = qkv.split(self.n_embd, dim=2)
        # Reshape into multiple heads
        Q = Q.view(B, T, self.n_heads, self.head_size).transpose(1, 2)
        K = K.view(B, T, self.n_heads, self.head_size).transpose(1, 2)
        V = V.view(B, T, self.n_heads, self.head_size).transpose(1, 2)
        # Attention scores
        scores  = (Q @ K.transpose(-2, -1)) / math.sqrt(self.head_size)
        scores  = scores.masked_fill(self.mask[:,:,:T,:T] == 0, float('-inf'))
        weights = F.softmax(scores, dim=-1)
        weights = self.dropout(weights)
        # Weighted sum of values
        out = weights @ V
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(out)

# --------------------------------------------------
# COMPONENT 2: Feed Forward
# --------------------------------------------------
class FeedForward(nn.Module):
    """Process information: expand 4x, then compress."""
    def __init__(self, config):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(config.n_embd, 4*config.n_embd),
            nn.GELU(),
            nn.Linear(4*config.n_embd, config.n_embd),
            nn.Dropout(config.dropout),
        )
    def forward(self, x):
        return self.net(x)

# --------------------------------------------------
# COMPONENT 3: Transformer Block
# --------------------------------------------------
class TransformerBlock(nn.Module):
    """One block: Attention + FeedForward with residuals."""
    def __init__(self, config):
        super().__init__()
        self.ln1       = nn.LayerNorm(config.n_embd)
        self.attention = SelfAttention(config)
        self.ln2       = nn.LayerNorm(config.n_embd)
        self.ffn       = FeedForward(config)
    def forward(self, x):
        # Residual connections (x + ...)
        x = x + self.attention(self.ln1(x))
        x = x + self.ffn(self.ln2(x))
        return x

# --------------------------------------------------
# COMPONENT 4: Complete Model
# --------------------------------------------------
class SLM(nn.Module):
    """Complete Small Language Model."""
    def __init__(self, config):
        super().__init__()
        self.config = config
        self.transformer = nn.ModuleDict({
            'token_emb': nn.Embedding(config.vocab_size, config.n_embd),
            'pos_emb'  : nn.Embedding(config.block_size, config.n_embd),
            'dropout'  : nn.Dropout(config.dropout),
            'blocks'   : nn.ModuleList([
                TransformerBlock(config) for _ in range(config.n_layers)
            ]),
            'ln_final' : nn.LayerNorm(config.n_embd),
        })
        self.lm_head = nn.Linear(config.n_embd, config.vocab_size, bias=False)
        # Weight tying — share token embedding with output head
        self.transformer['token_emb'].weight = self.lm_head.weight
        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx, targets=None):
        B, T    = idx.shape
        dev     = idx.device
        tok_emb = self.transformer['token_emb'](idx)
        pos     = torch.arange(T, device=dev)
        pos_emb = self.transformer['pos_emb'](pos)
        x       = self.transformer['dropout'](tok_emb + pos_emb)
        for block in self.transformer['blocks']:
            x = block(x)
        x       = self.transformer['ln_final'](x)
        logits  = self.lm_head(x)
        loss    = None
        if targets is not None:
            B, T, V = logits.shape
            loss = F.cross_entropy(
                logits.view(B*T, V), targets.view(B*T)
            )
        return logits, loss

    def count_parameters(self):
        return sum(p.numel() for p in self.parameters())

# --------------------------------------------------
# CREATE THE MODEL
# --------------------------------------------------
print("="*55)
print("  CREATING MODEL")
print("="*55)
print()

model = SLM(config).to(device)
total = model.count_parameters()

print(f"  ✅ Model created on {device.upper()}")
print(f"  Total parameters : {total:,} ({total/1e6:.1f}M)")
print()

# Quick forward pass test
print("  Testing forward pass...")
dummy_x = torch.randint(0, 100, (2, 32)).to(device)
dummy_y = torch.randint(0, 100, (2, 32)).to(device)
logits, loss = model(dummy_x, dummy_y)
print(f"  ✅ Forward pass works!")
print(f"     Output shape : {logits.shape}")
print(f"     Initial loss : {loss.item():.4f}")
print(f"     (random model loss ≈ {math.log(config.vocab_size):.2f})")
del dummy_x, dummy_y, logits, loss
torch.cuda.empty_cache()

print()
print("="*55)
print("✅ CELL 9 COMPLETE — Model Built!")
print("   Next → CELL 10: Test untrained output")
print("="*55)
```

```python
# CELL 11: Train the Model
# ==========================
import torch
import numpy as np
import os, time, math

# --------------------------------------------------
# CONFIG
# --------------------------------------------------
TOKENS_FILE    = "/kaggle/working/tokens.bin"
CHECKPOINT_DIR = "/kaggle/working/checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

BATCH_SIZE   = 32
BLOCK_SIZE   = config.block_size   # 256
MAX_STEPS    = 25000               # ~2-3 hours on T4
MAX_LR       = 3e-4
MIN_LR       = 3e-5
WARMUP_STEPS = 500
EVAL_EVERY   = 500
SAVE_EVERY   = 2500
TRAIN_SPLIT  = 0.95

print("="*55)
print("  TRAINING CONFIGURATION")
print("="*55)
print(f"  Batch size  : {BATCH_SIZE}")
print(f"  Block size  : {BLOCK_SIZE}")
print(f"  Max steps   : {MAX_STEPS:,}")
print(f"  Max LR      : {MAX_LR}")
print(f"  Tokens/step : {BATCH_SIZE*BLOCK_SIZE:,}")
print()

# --------------------------------------------------
# LOAD TOKENS
# --------------------------------------------------
print("📂 Loading tokens...")
tokens     = np.memmap(TOKENS_FILE, dtype=np.uint16, mode='r')
split_idx  = int(TRAIN_SPLIT * len(tokens))
train_data = tokens[:split_idx]
val_data   = tokens[split_idx:]
print(f"   Train : {len(train_data):,} tokens")
print(f"   Val   : {len(val_data):,} tokens")
print()

# --------------------------------------------------
# BATCH GENERATOR
# --------------------------------------------------
def get_batch(split='train'):
    data   = train_data if split == 'train' else val_data
    starts = torch.randint(len(data) - BLOCK_SIZE, (BATCH_SIZE,))
    x = torch.stack([
        torch.from_numpy(data[s:s+BLOCK_SIZE].astype(np.int64))
        for s in starts
    ])
    y = torch.stack([
        torch.from_numpy(data[s+1:s+BLOCK_SIZE+1].astype(np.int64))
        for s in starts
    ])
    return x.to(device), y.to(device)

# --------------------------------------------------
# LEARNING RATE SCHEDULE
# --------------------------------------------------
def get_lr(step):
    if step < WARMUP_STEPS:
        return MAX_LR * (step + 1) / WARMUP_STEPS
    if step > MAX_STEPS:
        return MIN_LR
    ratio = (step - WARMUP_STEPS) / (MAX_STEPS - WARMUP_STEPS)
    coeff = 0.5 * (1.0 + math.cos(math.pi * ratio))
    return MIN_LR + coeff * (MAX_LR - MIN_LR)

# --------------------------------------------------
# OPTIMIZER
# --------------------------------------------------
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=MAX_LR,
    betas=(0.9, 0.95),
    weight_decay=0.1
)
print("✅ Optimizer ready: AdamW")
print()

# --------------------------------------------------
# EVALUATION
# --------------------------------------------------
@torch.no_grad()
def evaluate(num_batches=20):
    model.eval()
    losses = []
    for _ in range(num_batches):
        x, y    = get_batch('val')
        _, loss = model(x, y)
        losses.append(loss.item())
    model.train()
    return sum(losses) / len(losses)

# --------------------------------------------------
# TRAINING LOOP
# --------------------------------------------------
print("="*55)
print("  TRAINING STARTED!")
print("="*55)
print()
print("  Step   | Loss   | LR       | Time/step")
print("  " + "-"*45)

model.train()
start_time  = time.time()
best_val    = float('inf')
recent_loss = []

for step in range(MAX_STEPS):
    # Set LR
    lr = get_lr(step)
    for pg in optimizer.param_groups:
        pg['lr'] = lr

    # Forward
    x, y = get_batch('train')
    optimizer.zero_grad()
    logits, loss = model(x, y)

    # Backward + update
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
    optimizer.step()

    recent_loss.append(loss.item())

    # Progress every 100 steps
    if step % 100 == 0:
        avg     = sum(recent_loss[-100:]) / min(len(recent_loss), 100)
        elapsed = time.time() - start_time
        ms      = elapsed / (step+1) * 1000
        print(f"  {step:>6} | {avg:.4f} | {lr:.2e} | {ms:.0f}ms")

    # Evaluate
    if step % EVAL_EVERY == 0 and step > 0:
        val  = evaluate()
        mark = " ← BEST!" if val < best_val else ""
        best_val = min(best_val, val)
        print(f"  {'─'*45}")
        print(f"  EVAL step {step}: val_loss={val:.4f}{mark}")
        print(f"  {'─'*45}")

    # Save checkpoint
    if step % SAVE_EVERY == 0 and step > 0:
        path = f"{CHECKPOINT_DIR}/ckpt_{step}.pt"
        torch.save({
            'step': step,
            'model_state': model.state_dict(),
            'config': config,
        }, path)
        print(f"  💾 Checkpoint saved: ckpt_{step}.pt")

# Final save
total_time = time.time() - start_time
torch.save({
    'step': MAX_STEPS,
    'model_state': model.state_dict(),
    'config': config,
}, "/kaggle/working/slm_final.pt")

print()
print("="*55)
print("🎉 TRAINING COMPLETE!")
print("="*55)
print(f"  Total steps   : {MAX_STEPS:,}")
print(f"  Total time    : {total_time/3600:.1f} hours")
print(f"  Best val loss : {best_val:.4f}")
print(f"  Final model   : slm_final.pt ✅")
print()
print("  Next → CELL 12: Chat with your model! 💬")
print("="*55)
```

```python
# CELL 12: Chat With Your Trained Model
# =======================================
# The demo moment — ask identity questions
# and see if the model learned WHO IT IS

import torch
import torch.nn.functional as F

def chat(question, max_new_tokens=150, temperature=0.7):
    """
    Ask the model a question, get its answer.
    Uses the same ### format from training.
    """
    model.eval()

    # Format prompt exactly like training data
    prompt = (
        f"### System:\n"
        f"You are a helpful AI assistant.\n\n"
        f"### Question:\n{question}\n\n"
        f"### Answer:\n"
    )

    input_ids = tokenizer.encode(prompt)
    input_ids = torch.tensor(
        input_ids, dtype=torch.long
    ).unsqueeze(0).to(device)

    generated = input_ids.clone()

    with torch.no_grad():
        for _ in range(max_new_tokens):
            inp = generated[:, -config.block_size:]
            logits, _ = model(inp)
            logits = logits[:, -1, :] / temperature

            # Top-k sampling: only consider top 50 tokens
            # This reduces random gibberish
            v, _ = torch.topk(logits, 50)
            logits[logits < v[:, [-1]]] = float('-inf')

            probs = F.softmax(logits, dim=-1)
            next_token = torch.multinomial(probs, num_samples=1)

            # Stop if model generates the separator (end of answer)
            generated = torch.cat([generated, next_token], dim=1)

            # Stop at next "###" (end of this answer)
            text_so_far = tokenizer.decode(
                generated[0][len(input_ids[0]):].tolist()
            )
            if "###" in text_so_far or "====" in text_so_far:
                break

    # Extract just the answer
    answer = tokenizer.decode(
        generated[0][len(input_ids[0]):].tolist()
    )
    # Clean up — stop at any separator
    for stop in ["###", "===="]:
        if stop in answer:
            answer = answer.split(stop)[0]

    return answer.strip()


# --------------------------------------------------
# THE BIG TEST — Identity Questions
# --------------------------------------------------
print("="*55)
print("  CHATTING WITH YOUR TRAINED MODEL")
print("="*55)

identity_questions = [
    "Who are you?",
    "What are you?",
    "What can you do?",
    "What are your limitations?",
    "Tell me about yourself.",
]

for q in identity_questions:
    print(f"\n{'─'*55}")
    print(f"❓ Q: {q}")
    print(f"{'─'*55}")
    answer = chat(q, max_new_tokens=150, temperature=0.7)
    print(f"🤖 A: {answer}")

print()
print("="*55)
print("  Compare this to the GARBAGE from Cell 10!")
print("  THAT is the power of training. 🎉")
print("="*55)
```
